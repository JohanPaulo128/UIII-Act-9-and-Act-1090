```python
# Listas independientes para cumplir con las reglas
nombres_enfermeros = []
años_experiencia = []

print("--- Sistema de Gestión de Enfermería con IA ---")

while True:
    # Interfaz de Menú
    print("\n===============================")
    print("      GESTIÓN DE PERSONAL")
    print("===============================")
    print("1. Registrar Enfermero")
    print("2. Mostrar Plantilla")
    print("3. Analizar Disponibilidad (IA)")
    print("4. Salir")
    print("===============================")

    opcion = input("Selecciona una opción (1-4): ")

    if opcion == "1":
        print("\n--- REGISTRO DE PERSONAL ---")
        nombre = input("Nombre del enfermero/a: ")
        # Convertimos a entero para poder comparar números después
        exp = int(input("Años de experiencia: "))

        nombres_enfermeros.append(nombre)
        años_experiencia.append(exp)
        print(">> Personal registrado correctamente.")

    elif opcion == "2":
        print("\n--- PLANTILLA ACTUAL ---")
        indice = 0
        if len(nombres_enfermeros) == 0:
            print("No hay personal registrado.")
        else:
            # Usamos while en lugar de for
            while indice < len(nombres_enfermeros):
                print("Enfermero:", nombres_enfermeros[indice], "| Experiencia:", años_experiencia[indice], "años")
                indice = indice + 1

    elif opcion == "3":
        print("\n--- ANÁLISIS DE ASIGNACIÓN IA ---")
        indice = 0
        if len(nombres_enfermeros) == 0:
            print("No hay datos para analizar.")
        else:
            while indice < len(nombres_enfermeros):
                nombre_actual = nombres_enfermeros[indice]
                exp_actual = años_experiencia[indice]

                # Lógica de IA con if / elif / else
                if exp_actual > 5:
                    print("IA:", nombre_actual, "-> Asignado a CUIDADOS INTENSIVOS (Alta Experiencia).")
                elif exp_actual >= 1:
                    print("IA:", nombre_actual, "-> Asignado a PLANTA GENERAL.")
                else:
                    print("IA:", nombre_actual, "-> Asignado a ACOMPAÑAMIENTO (Principiante).")

                indice = indice + 1

    elif opcion == "4":
        print("\nSaliendo del sistema de gestión... ¡Hasta pronto!")
        break

    else:
        print("\n[!] Opción no válida. Intenta de nuevo.")
```
