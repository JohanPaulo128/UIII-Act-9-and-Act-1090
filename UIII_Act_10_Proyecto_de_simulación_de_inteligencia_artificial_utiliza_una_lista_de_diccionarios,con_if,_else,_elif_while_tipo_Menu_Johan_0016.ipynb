```python
# Listas para guardar la información por separado (sin diccionarios ni objetos)
nombres = []
sintomas = []

print("--- Sistema de Triaje IA para Enfermería ---")

while True:
    print("\n===============================")
    print("       SISTEMA MEDICO IA")
    print("===============================")
    print("1. Registrar Paciente")
    print("2. Ver Lista de Pacientes")
    print("3. Diagnóstico IA (Analizar)")
    print("4. Salir")
    print("===============================")

    opcion = input("Selecciona una opción (1-4): ")

    if opcion == "1":
        print("\n--- NUEVO REGISTRO ---")
        nombre = input("Nombre del paciente: ")
        sintoma = input("Síntoma (fiebre / dolor / tos): ").lower()

        # Guardamos en las listas
        nombres.append(nombre)
        sintomas.append(sintoma)
        print("¡Paciente registrado con éxito!")

    elif opcion == "2":
        print("\n--- LISTA DE PACIENTES ACTUALES ---")
        # Usamos un while en lugar de un for
        i = 0
        if len(nombres) == 0:
            print("No hay pacientes en la lista.")
        else:
            while i < len(nombres):
                print("Paciente:", nombres[i], "| Síntoma:", sintomas[i])
                i = i + 1

    elif opcion == "3":
        print("\n--- ANÁLISIS DE DIAGNÓSTICO IA ---")
        # Usamos un while para analizar cada paciente
        i = 0
        if len(nombres) == 0:
            print("No hay pacientes para analizar.")
        else:
            while i < len(nombres):
                paciente_actual = nombres[i]
                sintoma_actual = sintomas[i]

                if sintoma_actual == "fiebre":
                    print("IA:", paciente_actual, "-> Alerta: Posible infección. Prioridad ALTA.")
                elif sintoma_actual == "tos":
                    print("IA:", paciente_actual, "-> Revisión pulmonar necesaria. Prioridad MEDIA.")
                elif sintoma_actual == "dolor":
                    print("IA:", paciente_actual, "-> Administrar analgésico. Prioridad BAJA.")
                else:
                    print("IA:", paciente_actual, "-> Síntoma desconocido. Bajo observación.")

                i = i + 1

    elif opcion == "4":
        print("\nCerrando sistema de triaje... ¡Buen turno, enfermero/a!")
        break

    else:
        print("\n[!] Opción inválida. Intenta de nuevo.")
```
