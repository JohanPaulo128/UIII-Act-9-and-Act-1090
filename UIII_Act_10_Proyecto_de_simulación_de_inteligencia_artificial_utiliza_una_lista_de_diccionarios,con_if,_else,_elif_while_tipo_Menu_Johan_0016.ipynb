```python
enfermeros = [] # Lista vacía
losenfermero = {}  # Diccionario vacío

while True:
    print("\n================================")
    print("   SISTEMA HOSPITALARIO IA   ")
    print("================================")
    print("1. Registrar Enfermero")
    print("2. Ver Plantilla Completa")
    print("3. IA: Análisis de Roles")
    print("4. Salir")
    print("================================")

    opcion = input("Selecciona una opción (1-4): ")

    if opcion == "1":
        nombre = input("Nombre del enfermero: ")
        especialidad = input("Especialidad (uci/piso/urgencias): ").lower()
        enfermero = {"nombre": nombre, "area": especialidad}
        enfermeros.append(enfermero)
        print("¡Datos guardados!")

    elif opcion == "2":
        for e in enfermeros:
            print(f"Personal: {e['nombre']} | Área: {e['area']}")

    elif opcion == "3":
        for e in enfermeros:
            if e["area"] == "uci":
                print(f"IA: {e['nombre']} requiere monitoreo constante.")
            elif e["area"] == "urgencias":
                print(f"IA: {e['nombre']} asignado a clasificación rápida.")
            else:
                print(f"IA: {e['nombre']} asignado a cuidados generales.")

    elif opcion == "4":
        print("Apagando sistema...")
        break
    else:
        print("Opción no válida.")
```
