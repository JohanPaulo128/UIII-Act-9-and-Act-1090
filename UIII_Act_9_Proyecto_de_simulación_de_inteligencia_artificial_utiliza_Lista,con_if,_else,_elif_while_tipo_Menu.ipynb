## Clase = Enfermero

```python
pacientes = []
opcion = ""

while opcion != "3":
    print("\n--- 🏥 Sistema de Enfermería Virtual ---")
    print("1. Registrar llegada de paciente")
    print("2. Ver lista de espera")
    print("3. Salir")
    opcion = input("Seleccione una opción: ")

    if opcion == "1":
        nombre = input("Nombre del paciente: ")
        pacientes.append(nombre)
        print("¡Paciente registrado!")
    elif opcion == "2":
        if pacientes == []:
            print("La lista está vacía. No hay pacientes.")
        else:
            print("Pacientes en espera:", pacientes)
    elif opcion == "3":
        print("Cerrando sistema médico...")
    else:
        print("Opción no válida, intenta de nuevo.")
```
